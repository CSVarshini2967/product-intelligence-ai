```python
import pandas as pd

# Change the path to your dataset
df = pd.read_csv("sample_dataset.csv")

print("="*60)
print("Dataset Shape")
print("="*60)
print(df.shape)

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names")
print("="*60)
print(df.columns.tolist())

print("\nFirst 5 Rows")
print("="*60)
print(df.head())

print("\nLast 5 Rows")
print("="*60)
print(df.tail())
```

```python
print("="*60)
print("Data Types")
print("="*60)

print(df.dtypes)
```

```python
print("="*60)
print("Missing Values")
print("="*60)

missing = df.isnull().sum()

missing_percent = (missing / len(df)) * 100

summary = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_percent.round(2)
})

print(summary.sort_values("Missing %", ascending=False))
```

```python
print("="*60)
print("Duplicate Rows")
print("="*60)

print(df.duplicated().sum())
```

```python
for col in df.columns:
    print("\n")
    print("="*70)
    print(col)
    print("="*70)

    print("Unique Values:", df[col].nunique())

    print(df[col].value_counts(dropna=False).head(10))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,6))

plt.imshow(df.isnull(), aspect="auto")

plt.xlabel("Columns")
plt.ylabel("Rows")

plt.title("Missing Values")

plt.colorbar()

plt.show()
```

```python
text_cols = df.select_dtypes(include="object").columns

for col in text_cols:
    print("\n")
    print(col)

    print(df[col].astype(str).str.len().describe())
```

```python
for col in df.columns:

    print("\n")
    print("="*80)
    print(col)
    print("="*80)

    print(df[col].describe(include="all"))
```

```python
report = pd.DataFrame({
    "Column": df.columns,
    "Datatype": df.dtypes.values,
    "Missing": df.isnull().sum().values,
    "Unique": df.nunique().values
})

report.to_csv("dataset_report.csv", index=False)

print("Dataset report saved!")
```
